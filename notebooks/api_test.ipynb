```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv('OPENWEATHER_API_KEY')

CITY = "Konya,TR"

# Sıcaklıkları Celsius almak için units=metric
URL = f"https://api.openweathermap.org/data/2.5/weather?q={CITY}&appid={API_KEY}&units=metric&lang=tr"

# API'ye isteği gönder
response = requests.get(URL)
data = response.json()

if response.status_code == 200:
    print(f"Şehir: {data['name']}")
    print(f"Sıcaklık (Dış Ortam): {data['main']['temp']} °C")
    print(f"Bağıl Nem (Dış Ortam): %{data['main']['humidity']}")
    print(f"Hava Durumu: {data['weather'][0]['description'].capitalize()}")
    print(f"Atmosferik Basınç: {data['main']['pressure']} hPa")
else:
    print("API Hatası:", data['message'])
```

```python
# Gelecek 24 saatin (3'er saatlik aralıklarla) tahminini çek
URL_FORECAST = f"https://api.openweathermap.org/data/2.5/forecast?q={CITY}&appid={API_KEY}&units=metric&lang=tr"

response_forecast = requests.get(URL_FORECAST)
data_forecast = response_forecast.json()

if response_forecast.status_code == 200:
    print(f"--- {data_forecast['city']['name']} Gelecek 24 Saatlik Zaman Serisi Tahmini ---")

    for zaman in data_forecast['list'][:8]:
        saat = zaman['dt_txt']
        sicaklik = zaman['main']['temp']
        nem = zaman['main']['humidity']
        durum = zaman['weather'][0]['description']
        print(f"Zaman: {saat} | Sıcaklık: {sicaklik}°C | Nem: %{nem} | Durum: {durum}")
else:
    print("API Hatası:", data_forecast['message'])
```
